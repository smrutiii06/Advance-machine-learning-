```python
import numpy as np
import pandas as pd

rainfall = ["high","high","high","low","low","low"]
soiltype = ["loamy","sandy","clay","loamy","sandy","clay"]
yeild = ["good","good","good","poor","poor","good",]

def rone(df):
    col = list(df.columns)
    target = col[-1]
    length = len(df[target])
    resPer = None
    res = ""
    for i in col[:-1]:
        currError = 0
        currCol = set(df[i])
        for j in currCol:
            temp = df[df[i] == j][target].value_counts()

            if len(temp) > 0:
                currError += temp.sum() - temp.max()
        if resPer == None:
            resPer = currError/length
            res = i
        elif currError/length < resPer:
            resPer = currError/length
            res = i
    print(f"The feature {res} is selected with error rate {resPer*100:.4f}%")
data = {"rainfall":rainfall, "soiltype":soiltype, "yeild":yeild}
df = pd.DataFrame(data)

rone(df)
```
